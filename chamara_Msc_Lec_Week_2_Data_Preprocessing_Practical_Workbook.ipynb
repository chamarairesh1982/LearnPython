## 1 Data Profiling
Data profiling is about measuring statistical information about data. It helps you to get a better insight of the quality of data as well as the structure, pattern and anomalies may exist within it.
* Data profiling can be started with identifying data types within a dataset.
* Next finding the missing values, duplicate values within a dataset.
* Determining uniqueness (the number of rows in a column with unique values) is important.
  - High uniqueness (few rows with duplicate values) can make it easier to build ML models, but it may also reduce the model's accuracy. This is because the model may not have enough data to learn the relationships between the features and the target variable.
  - Low uniqueness (many rows with duplicate values) can make it more difficult to build ML models, but it may also increase the model's accuracy.
*  The cardinality of data samples per feature (the number of distinct values in a column) is also important to consider.
    - High cardinality (many distinct values) can make it difficult to build ML models that accurately predict the values of the feature, as the model may struggle to learn the relationships between the feature and the target variable.
    - Low cardinality (few distinct values) can make it easier to build models, but it may also reduce their accuracy.
* Calculating statistical measures.
    - Count: number of non-null values in each column. If the count is significantly less than the total number of rows in the DataFrame, it suggests the presence of missing values.
    - Mean: mean (average) gives you an idea of the central tendency of the data. If the mean is close to the median (50th percentile), it suggests a symmetric distribution. However, if the mean is significantly different from the median, it may indicate a skewed distribution.
    - Standard Deviation: spread or dispersion of the data. A higher standard deviation suggests a wider distribution, while a lower standard deviation indicates a narrower distribution.
    - Minimum and Maximum: The minimum and maximum values provide insights into the range of the data. They can help you identify outliers or extreme values that may influence the shape of the distribution.
   - Quartiles (25th, 50th, and 75th percentiles): The quartiles divide the data into four equal parts. The 50th percentile (median) represents the value below which 50% of the data falls. Comparing the quartiles can give you an indication of the skewness of the distribution. If the 25th percentile is close to the 50th percentile, it suggests a symmetrical distribution. Conversely, if the 25th percentile is significantly smaller or larger than the 50th percentile, it indicates a skewed distribution.

## 3 Data Cleaning

```python
# Remove Duplicates
df=df.drop_duplicates(); df
print(" ** Results after removal:",df.duplicated().sum())
```

```python
# Handle Missing Values
# Delete rows having missing values (The simplest way), filling missing values with a default value or appropriate average value (mean, median, or mode).
# You can also use a model like Gaussian to find the best value to be substituted with the missing value.
df['Age']=df['Age'].fillna(df['Age'].median())
df['Attendance']=df['Attendance'].fillna(df['Attendance'].mean())
df['Assignment']=df['Assignment'].fillna(df['Assignment'].mean())
df
```

```python
# Fix Inconsistent Values
df['City']=df['City'].str.title()
df['City'].unique()
```

```python
# Detect Outliers
Q1=df['Age'].quantile(.25);Q3=df['Age'].quantile(.75);IQR=Q3-Q1
lower=Q1-1.5*IQR; upper=Q3+1.5*IQR
df[(df['Age']<lower)|(df['Age']>upper)]
```

```python
# Label Encoding: Convert categorical text data into numerical integers so that machine learning algorithms can process it
le=LabelEncoder()
for col in ['Gender','Department','City','Placement']:
    df[col]=le.fit_transform(df[col])
df
```

## 6 Data Transformation

```python
# Normalization (Min-Max Scaling)
m=MinMaxScaler()
cols=['Age','Attendance','Assignment','Exam']
df[cols]=m.fit_transform(df[cols]); df
```

```python
#  Standardization (Z-score Scaling)
s=StandardScaler()
pd.DataFrame(s.fit_transform(df[cols]),columns=[i+'_std' for i in cols])
```

```python
# Discretization
df['Exam_Level']=pd.cut(df['Exam'],bins=3,labels=['Low','Medium','High'])
df[['Exam','Exam_Level']]
```

# 7 Data Reduction
