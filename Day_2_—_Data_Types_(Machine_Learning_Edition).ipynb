```python
dataset_name = "Cyber Attack Dataset"

total_records = 10000
missing_records = 250
duplicate_records = 100

valid_records = total_records - missing_records - duplicate_records

data_quality_percentage = valid_records / total_records * 100

dataset_ready = data_quality_percentage >= 95

print("================================")
print("Data Quality Report")
print("================================")
print("Dataset:", dataset_name)
print("Total Records:", total_records)
print("Missing Records:", missing_records)
print("Duplicate Records:", duplicate_records)
print("Valid Records:", valid_records)
print(
    "Data Quality:",
    round(data_quality_percentage, 2),
    "%"
)
print("Dataset Ready:", dataset_ready)
print("================================")
```

```python
# Project information
project_name = "Network Intrusion Detection"
dataset_name = "CICIDS2017"
algorithm = "Random Forest"

# Dataset information
number_of_samples = 225745
number_of_features = 78

# Training information
training_time_hours = 2.35
epochs = 100
learning_rate = 0.001

# Performance metrics
accuracy = 98.72
precision = 98.11
recall = 97.85

# Calculate F1 score
f1_score = 2 * (precision * recall) / (precision + recall)

# Model status
training_completed = True
model_saved = True

print("=========================================")
print("      Machine Learning Training Report")
print("=========================================")

print("Project           :", project_name)
print()

print("Dataset           :", dataset_name)
print("Algorithm         :", algorithm)
print()

print("Samples           :", number_of_samples)
print("Features          :", number_of_features)
print()

print("Epochs            :", epochs)
print("Learning Rate     :", learning_rate)
print("Training Time     :", training_time_hours, "Hours")
print()

print("Accuracy          :", accuracy, "%")
print("Precision         :", precision, "%")
print("Recall            :", recall, "%")
print("F1 Score          :", round(f1_score, 2), "%")
print()

print("Training Completed:", training_completed)
print("Model Saved       :", model_saved)

print("=========================================")
```

```python
print("Enter Machine Learning Experiment Details")
print("-----------------------------------------")

project_name = input("Project name: ")
dataset_name = input("Dataset name: ")
algorithm = input("Algorithm: ")

samples = int(input("Number of samples: "))
features = int(input("Number of features: "))

training_time = float(input("Training time in hours: "))
accuracy = float(input("Accuracy: "))
precision = float(input("Precision: "))
recall = float(input("Recall: "))

training_status_text = input(
    "Training completed? True or False: "
)

training_completed = (
    training_status_text.strip().lower() == "true"
)

f1_score = 2 * (precision * recall) / (precision + recall)

model_ready = (
    training_completed
    and accuracy >= 90
    and f1_score >= 90
)

print("\n=========================================")
print("      Machine Learning Training Report")
print("=========================================")
print(f"Project           : {project_name}")
print(f"Dataset           : {dataset_name}")
print(f"Algorithm         : {algorithm}")
print()
print(f"Samples           : {samples:,}")
print(f"Features          : {features}")
print(f"Training Time     : {training_time:.2f} hours")
print()
print(f"Accuracy          : {accuracy:.2f}%")
print(f"Precision         : {precision:.2f}%")
print(f"Recall            : {recall:.2f}%")
print(f"F1 Score          : {f1_score:.2f}%")
print()
print(f"Training Completed: {training_completed}")
print(f"Model Ready       : {model_ready}")
print("=========================================")
```
